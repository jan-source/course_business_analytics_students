## 1. Variables and Data Types <a id="variables"></a>

Python has several built-in data types. Let's explore the most common ones:

```python
# Numbers
age = 25
height = 5.9
temperature = -10.5

# Strings
name = "Alice"
city = 'New York'
message = """This is a
multi-line string"""

# Boolean
is_student = True
is_working = False

# Print values and types
print(f"Age: {age}, Type: {type(age)}")
print(f"Height: {height}, Type: {type(height)}")
print(f"Name: {name}, Type: {type(name)}")
print(f"Is student: {is_student}, Type: {type(is_student)}")
```

## 2. Basic Operations <a id="operations"></a>

Let's perform some basic mathematical and string operations:

```python
# Mathematical operations
a = 10
b = 3

print(f"Addition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")
print(f"Integer Division: {a} // {b} = {a // b}")
print(f"Modulo: {a} % {b} = {a % b}")
print(f"Exponentiation: {a} ** {b} = {a ** b}")
```

```python
# String operations
first_name = "John"
last_name = "Doe"

# Concatenation
full_name = first_name + " " + last_name
print(f"Full name: {full_name}")

# String methods
print(f"Uppercase: {full_name.upper()}")
print(f"Lowercase: {full_name.lower()}")
print(f"Length: {len(full_name)}")
print(f"Replace: {full_name.replace('John', 'Jane')}")
```

## 3. Lists and Tuples <a id="lists"></a>

Lists are mutable collections, while tuples are immutable:

```python
# Lists
fruits = ["apple", "banana", "orange", "grape"]
numbers = [1, 2, 3, 4, 5]
mixed = ["hello", 42, 3.14, True]

print(f"Fruits: {fruits}")
print(f"First fruit: {fruits[0]}")
print(f"Last fruit: {fruits[-1]}")
print(f"First three fruits: {fruits[:3]}")

# List methods
fruits.append("mango")
print(f"After adding mango: {fruits}")

fruits.remove("banana")
print(f"After removing banana: {fruits}")

print(f"Length of fruits list: {len(fruits)}")
```

```python
# Tuples
coordinates = (10, 20)
colors = ("red", "green", "blue")

print(f"Coordinates: {coordinates}")
print(f"X coordinate: {coordinates[0]}")
print(f"Y coordinate: {coordinates[1]}")

# Tuples are immutable - this would cause an error:
# coordinates[0] = 15  # Uncomment to see the error
```

## 4. Dictionaries <a id="dictionaries"></a>

Dictionaries store key-value pairs:

```python
# Creating dictionaries
student = {
    "name": "Alice",
    "age": 22,
    "major": "Computer Science",
    "grades": [85, 92, 78, 95]
}

print(f"Student info: {student}")
print(f"Name: {student['name']}")
print(f"Age: {student['age']}")
print(f"Grades: {student['grades']}")

# Adding new key-value pairs
student["email"] = "alice@example.com"
print(f"After adding email: {student}")

# Dictionary methods
print(f"Keys: {student.keys()}")
print(f"Values: {student.values()}")
print(f"Items: {student.items()}")
```

## 5. Control Flow <a id="control"></a>

Control flow statements help you make decisions and repeat actions:

```python
# If-elif-else statements
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
```

```python
# For loops
print("Iterating through a list:")
fruits = ["apple", "banana", "orange"]
for fruit in fruits:
    print(f"I like {fruit}")

print("\nIterating through a range:")
for i in range(5):
    print(f"Number: {i}")

print("\nIterating through a dictionary:")
student = {"name": "Alice", "age": 22, "major": "CS"}
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# While loops
print("Countdown:")
count = 5
while count > 0:
    print(count)
    count -= 1
print("Blast off!")
```

## 6. Functions <a id="functions"></a>

Functions help organize and reuse code:

```python
# Simple function
def greet(name):
    return f"Hello, {name}!"

# Function with multiple parameters
def calculate_area(length, width):
    return length * width

# Function with default parameter
def power(base, exponent=2):
    return base ** exponent

# Using functions
print(greet("Alice"))
print(f"Area of 5x3 rectangle: {calculate_area(5, 3)}")
print(f"2 squared: {power(2)}")
print(f"2 cubed: {power(2, 3)}")
```

```python
# Function with list processing
def calculate_statistics(numbers):
    if not numbers:
        return None
    
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    minimum = min(numbers)
    maximum = max(numbers)
    
    return {
        "total": total,
        "count": count,
        "average": average,
        "minimum": minimum,
        "maximum": maximum
    }

# Test the function
test_numbers = [10, 25, 5, 30, 15, 20]
stats = calculate_statistics(test_numbers)
print(f"Numbers: {test_numbers}")
print(f"Statistics: {stats}")
```

## 7. Working with Libraries <a id="libraries"></a>

Python has a rich ecosystem of libraries. Let's explore some basic ones:

```python
# Math library
import math

print(f"Pi: {math.pi}")
print(f"Square root of 16: {math.sqrt(16)}")
print(f"Sine of 90 degrees: {math.sin(math.radians(90))}")
print(f"Ceiling of 4.2: {math.ceil(4.2)}")
print(f"Floor of 4.8: {math.floor(4.8)}")
```

```python
# Random library
import random

print(f"Random integer between 1 and 10: {random.randint(1, 10)}")
print(f"Random float between 0 and 1: {random.random()}")

# Random choice from a list
colors = ["red", "blue", "green", "yellow"]
print(f"Random color: {random.choice(colors)}")

# Shuffle a list
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(f"Shuffled numbers: {numbers}")
```

```python
# DateTime library
from datetime import datetime, date

now = datetime.now()
today = date.today()

print(f"Current date and time: {now}")
print(f"Today's date: {today}")
print(f"Formatted date: {now.strftime('%Y-%m-%d %H:%M:%S')}")
```
